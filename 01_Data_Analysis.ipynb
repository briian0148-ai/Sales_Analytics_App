### Tratamiento de registros duplicados

Después de revisar los registros duplicados, se identificaron 5,268 filas duplicadas de manera exacta, lo que representaba aproximadamente el 0.97% del dataset antes de su eliminación.

El análisis mostró que algunos registros aparecían repetidos múltiples veces dentro de la misma factura, llegando incluso a encontrarse casos de una misma transacción repetida 20 veces. Debido a que estas filas eran idénticas en todas sus columnas, mantenerlas podría generar una sobreestimación de las ventas.

Por esta razón, se eliminaron únicamente los duplicados exactos utilizando `drop_duplicates()`. Después del proceso, se verificó que no quedaran registros duplicados en el dataset.

```python
frecuencia_cliente = (
    df_clientes
    .groupby("CustomerID")["InvoiceDate"]
    .agg(
        PrimeraCompra="min",
        UltimaCompra="max",
        NumeroCompras="nunique"
    )
)

frecuencia_cliente.head()
```

```python
frecuencia_cliente["DiasActivo"] = (
    frecuencia_cliente["UltimaCompra"] -
    frecuencia_cliente["PrimeraCompra"]
).dt.days

frecuencia_cliente.head(10)
```

```python
frecuencia_cliente["DiasEntreCompras"] = (
    frecuencia_cliente["DiasActivo"] /
    (frecuencia_cliente["NumeroCompras"] - 1)
).where(
    frecuencia_cliente["NumeroCompras"] > 1
)

frecuencia_cliente.head(10)
```

```python
fecha_referencia = df_clientes["InvoiceDate"].max() + pd.Timedelta(days=1)

print("Fecha de referencia:", fecha_referencia)
```

```python
frecuencia_cliente["Recency"] = (
    fecha_referencia -
    frecuencia_cliente["UltimaCompra"]
).dt.days

frecuencia_cliente.head(10)
```

```python
rfm = pd.DataFrame({
    "Recency": frecuencia_cliente["Recency"],
    "Frequency": frecuencia_cliente["NumeroCompras"],
    "Monetary": gasto_cliente
})

rfm.head(10)
```

```python
rfm.describe()
```

```python
rfm['R_score'] = pd.qcut(
    rfm['Recency'],
    5,
    labels=[5, 4, 3, 2, 1]
)

rfm['F_score'] = pd.qcut(
    rfm['Frequency'].rank(method='first'),
    5,
    labels=[1, 2, 3, 4, 5]
)

rfm['M_score'] = pd.qcut(
    rfm['Monetary'].rank(method='first'),
    5,
    labels=[1, 2, 3, 4, 5]
)
```

```python
rfm.head(10)
```

```python
print("R_score:")
print(rfm['R_score'].value_counts().sort_index())

print("\nF_score:")
print(rfm['F_score'].value_counts().sort_index())

print("\nM_score:")
print(rfm['M_score'].value_counts().sort_index())
```

```python
rfm['RFM_Score'] = (
    rfm['R_score'].astype(str) +
    rfm['F_score'].astype(str) +
    rfm['M_score'].astype(str)
)

rfm.head(10)
```

```python
rfm['RFM_Total'] = (
    rfm['R_score'].astype(int) +
    rfm['F_score'].astype(int) +
    rfm['M_score'].astype(int)
)

rfm[['Recency', 'Frequency', 'Monetary',
     'R_score', 'F_score', 'M_score',
     'RFM_Score', 'RFM_Total']].head(10)
```
